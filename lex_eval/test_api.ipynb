```python
import requests
import json

def search_uk_legislation(query_text):
    """
    Sends a semantic search query to the Lex API and processes the response.
    """
    url = "https://lex.lab.i.ai.gov.uk/legislation/search"
    headers = {"Content-Type": "application/json"}
    payload = {"query": query_text, "include_text": False}

    try:
        # Sending the POST request
        response = requests.post(url, headers=headers, json=payload)
        
        # Checking if the request was successful
        response.raise_for_status()
        
        # Parsing the JSON response
        data = response.json()
        return data

    except requests.exceptions.RequestException as e:
        print(f"An error occurred: {e}")
        return None

def analyse_results(data):
    """
    Analyses and prints a summary of the API response.
    """
    if not data or 'results' not in data:
        print("No results found or invalid data format.")
        return

    total_results = data.get('total', 0)
    results = data.get('results', [])

    print(f"--- Search Analysis ---")
    print(f"Total Matches Found: {total_results}")
    print(f"Showing Top {len(results)} results:\n")

    # Table Header
    print(f"{'Score':<8} | {'Year':<6} | {'Title'}")
    print("-" * 60)

    for item in results:
        # Extracting key fields
        title = item.get('title', 'Unknown Title')
        year = item.get('year', 'N/A')
        score = item.get('score', 0.0)
        
        # Formatting output
        print(f"{score:<8.4f} | {year:<6} | {title}")

# --- Execution ---
search_query = "chair of food standards scotland appointment"
response_data = search_uk_legislation(search_query)

if response_data:
    analyse_results(response_data)
```

```python
import requests
import json

def search_within_legislation(legislation_uri, query_text):
    """
    Searches for specific sections using the full legislation URI.
    """
    url = "https://lex.lab.i.ai.gov.uk/legislation/section/search"
    headers = {"Content-Type": "application/json"}
    
    # Based on your output, the legislation_id is the full URL
    payload = {
        "legislation_id": legislation_uri,
        "query": query_text
    }

    try:
        response = requests.post(url, headers=headers, json=payload)
        response.raise_for_status()
        return response.json()
    except requests.exceptions.RequestException as e:
        print(f"Request failed: {e}")
        return None

def analyse_sections(data):
    """
    Processes the list of dictionaries returned by the API.
    """
    if not data or not isinstance(data, list):
        print("No results found or unexpected format.")
        return

    print(f"Found {len(data)} matching sections:\n")
    
    for item in data:
        # Extracting data using keys identified in your raw output
        number = item.get('number', 'N/A')
        title = item.get('title', 'Untitled')
        text_preview = item.get('text', '')[:150] # Snippet from the 'text' field
        
        print(f"--- Section {number}: {title} ---")
        print(f"Preview: {text_preview.strip()}...")
        print(f"Link: {item.get('uri')}")
        print("\n")

# --- Usage ---
# Use the full URI as seen in your 'legislation_id' field
full_uri = "ukpga/2018/12"
search_term = "controller"

results = search_within_legislation(full_uri, search_term)
analyse_sections(results)
```
